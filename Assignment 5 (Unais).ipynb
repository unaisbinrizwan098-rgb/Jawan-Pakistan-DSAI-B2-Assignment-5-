# Q1. Count vowels

```python
def count_vowels(text):
    vowels = set("aeiouAEIOU")
    return sum(1 for ch in text if ch in vowels)
```

```python
print(count_vowels("Welcome"))
```

# Q2. Display student info using keyword arguments


```python
def display_student_info(name, age, grade, school):
    print(f"Name: {name}")
    print(f"Age: {age}")
    print(f"Grade: {grade}")
    print(f"School: {school}")
```

```python
display_student_info(
    name="Javeria Unais",
    age=23,
    grade="Undergraduate",
    school="PECHS College"
)
```

# Q3. Display product prices using arbitrary keyword arguments


```python
def display_prices(product_name, **prices):
    print(f"Product: {product_name}")
    for store, price in prices.items():
        print(f"{store}: ${price}")
```

```python
display_prices("Laptop", Amazon=800, eBay=750, Walmart=780)
```

# Q4. Find maximum using arbitrary positional arguments


```python
def find_max(*numbers):
    if not numbers:
        return None
    return max(numbers)
```

```python
print(find_max(3, 7, 2, 9, 1))
```

# Q6. Rectangle area and perimeter using nested functions


```python
def calculate_rectangle(length, width):
    def calculate_area():
        return length * width

    def calculate_perimeter():
        return 2 * (length + width)

    return calculate_area(), calculate_perimeter()

area, perimeter = calculate_rectangle(4, 8)
```

```python
print(f"Area: {area}")
print(f"Perimeter: {perimeter}")
```

# Q8. Hotel bill using nested functions


```python
def hotel_bill(nights):
    rate_per_night = 8000

    def room_cost():
        return nights * rate_per_night

    def tax(cost):
        return cost * 0.10

    cost = room_cost()
    tax_amount = tax(cost)
    total_bill = cost + tax_amount

    return cost, tax_amount, total_bill

room_cost_value, tax_value, total_value = hotel_bill(3)
```

```python
print(f"Room Cost: {room_cost_value}")
print(f"Tax: {tax_value}")
print(f"Total Bill: {total_value}")
```

# Q9. User login validation using nested function


```python
def validate_login(username, password):
    stored_username = "Unais"
    stored_password = "unais123"

    def validate():
        return username == stored_username and password == stored_password

    if validate():
        return "Login successful"
    else:
        return "Invalid username or password"
```

```python
print(validate_login("Unais", "unais123"))  
print(validate_login("Unais", "wrong"))     
```

# Q10. Shopping bill using nested functions


```python
def shopping_bill(price, quantity, discount=0):
    def calculate_total():
        return price * quantity

    def apply_discount(total):
        return max(0, total - discount)

    total = calculate_total()
    return apply_discount(total)   
```

```python
print(f"Final Bill = {shopping_bill(100, 7)} PKR")    
print(f"Final Bill = {shopping_bill(100, 7, 100)} PKR")
```
